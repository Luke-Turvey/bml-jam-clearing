```python
from sklearn.linear_model import LinearRegression

def linreg(X,Y):
    A = np.vstack([X, np.ones(len(X))]).T
    m, c = np.linalg.lstsq(A, Y, rcond=None)[0]


    return m, c
```

```python


b=2
x = np.array(df.iloc[30:,0])
y = np.array(df.iloc[30:,1])

plt.grid(True, which="major", linestyle='-', linewidth=0.3, zorder=0)
plt.grid(True, which="minor", linestyle='-', linewidth=0.1, zorder=0)

plt.scatter(x=x,y=y,marker='x')

plt.xscale('log')
plt.yscale('log')

X,Y = np.log(x) / np.log(b) , np.log(y) / np.log(b)
m,c = linreg(X,Y)

slx = X[x % 3 == 2]
sly = Y[x % 3 == 2]
lx = X[x % 3 != 2]
ly = Y[x % 3 != 2]

slm, slc = linreg(slx, sly)
lm, lc = linreg(lx,ly)

print(f"y={round(slm,3)} x + {round(slc, 2)}")
print(f"y={round(lm,3)} x + {round(lc, 2)}")

plt.plot( 2**slx, 2**(slm*slx+slc), color = 'r', alpha = 0.2 )
plt.plot( 2**lx, 2**(lm*lx+lc), color = 'r', alpha = 0.2 )
plt.xlabel("$n$")
plt.ylabel("$T(G_n,n)$")
plt.savefig(f'log2_133333113_plot.png')
plt.show()

```
